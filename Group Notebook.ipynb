Austin Sheets - Property Condition Vs Price of Housing. Most used metrics were Property value and condition of housing.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

s = pd.read_csv("Housing Price - Austin Sheets/housing_by_neighborhood.csv")
```

```python
top = s.sort_values("housing_score", ascending=False).head(10).sort_values("housing_score")
top.plot.barh(x="NEIGHBORHOOD", y="housing_score", legend=False, figsize=(8, 5))
plt.xlabel("Housing score (0-100; higher = better quality for the price)")
plt.title("Top 10 neighborhoods by combined housing score")
plt.show()
```

Conclusion - Through my data searching I have found that when comparing quality of house to cost of house, the neighborhoods above score within the top 10. I believe my data supports this graph since Squirrel Hill North is often said to be one of the nicest neighborhoods in Pittsburgh.

**Christopher Lattari** - Crime Rate based on Population per Neighborhood

```python
crimeRate= pd.read_csv("CrimeRate_Neighborhood.csv", sep=",")
neighborhoods= crimeRate["Neighborhood"][:10]
cr= crimeRate["Crime Rate per 1000"][:10]

plt.barh(neighborhoods,cr, color='red')
plt.xlabel('Crime Rate per 1000')
plt.ylabel('Neighborhoods')
plt.title('Top 10 Neighborhoods By Lowest Crime Rate')
plt.show()
```

These results were found using two datasets. The first was a list of crimes reported and the neighborhoods they occurred in alongside some other information that I didn't have to use. The second dataset contained the population of each neighborhood. Using pandas to access both of these I found the number of crimes committed in each neighborhood which I was then able to divide by that neighborhoods population and multipy by 1000 to find the crime rate per 1000 people. 

**Conclusion**- I ultimately ended up with the list above as my top ten best neighborhoods with Swisshelm Park taking my number one spot as the best.
