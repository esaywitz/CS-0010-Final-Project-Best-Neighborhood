number of bus stops by city

```python
import pandas as pd
%matplotlib inline
import matplotlib.pyplot as plt

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None,
    "display.width", None,
    "display.max_colwidth", None,
):

    stops = pd.read_csv("bus-stops.tsv", sep="\t")

# count occurrences of number of bus stops per city
occur = stops.groupby(['hood']).size()
top5 = occur.sort_values(ascending=False).head(5)
print(top5)

hood_in_order = stops.sort_values(['hood'])
# get all cities together

hood_list = stops.groupby('hood')['route_code'].unique()

top5.plot(kind="bar", title="Top 5 Neighborhoods by bus stops", xlabel="Hood", ylabel="Count")
plt.show()

for hood in top5.index:
    print(hood)
    print(hood_list.get(hood).unique())
```
