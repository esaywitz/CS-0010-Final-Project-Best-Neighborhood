```python
import pandas as pd

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None,
    "display.width", None,
    "display.max_colwidth", None,
):

    stops = pd.read_csv("bus-stops.tsv", sep="\t")

# count occurrences a particular column
    occur = stops.groupby(['hood']).size()
    occur.sort_values()
    top5 = occur.sort_values(ascending=False).head(5)

# display occurrences of a particular column
    print(top5)

# print above threshold
    threshold = 20
```
